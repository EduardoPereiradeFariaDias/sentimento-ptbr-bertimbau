```python
from datasets import load_dataset

ds = load_dataset("ruanchaves/b2w-reviews01", revision="refs/convert/parquet", split="train")
print(ds.column_names)
print(ds[0])
```

```python
ds = ds.filter(lambda x: x["review_text"] and len(x["review_text"]) > 0)

def to_label(rating):
    return 0 if rating <= 2 else (1 if rating == 3 else 2)

ds = ds.map(lambda x: {"label": to_label(x["overall_rating"])})
ds = ds.shuffle(seed=42).select(range(30000))

split = ds.train_test_split(test_size=0.2, seed=42)
tmp = split["test"].train_test_split(test_size=0.5, seed=42)
train, val, test = split["train"], tmp["train"], tmp["test"]

print(len(train), len(val), len(test))
print(ds.to_pandas()["label"].value_counts())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, f1_score

vec = TfidfVectorizer(max_features=50000, ngram_range=(1, 2))
X_train = vec.fit_transform(train["review_text"])
X_test = vec.transform(test["review_text"])

clf = LogisticRegression(max_iter=1000)
clf.fit(X_train, train["label"])

preds_base = clf.predict(X_test)
print(classification_report(test["label"], preds_base,
      target_names=["negativo", "neutro", "positivo"]))
print("F1 macro baseline:", f1_score(test["label"], preds_base, average="macro"))
```

```python
import numpy as np
from transformers import (AutoTokenizer, AutoModelForSequenceClassification,
                          TrainingArguments, Trainer, DataCollatorWithPadding)
from sklearn.metrics import f1_score, accuracy_score

model_name = "neuralmind/bert-base-portuguese-cased"
tok = AutoTokenizer.from_pretrained(model_name)

def tokenize(batch):
    return tok(batch["review_text"], truncation=True, max_length=128)

train_t = train.map(tokenize, batched=True)
val_t = val.map(tokenize, batched=True)
test_t = test.map(tokenize, batched=True)

id2label = {0: "negativo", 1: "neutro", 2: "positivo"}
model = AutoModelForSequenceClassification.from_pretrained(
    model_name, num_labels=3, id2label=id2label,
    label2id={v: k for k, v in id2label.items()})

def metrics(p):
    preds = np.argmax(p.predictions, axis=1)
    return {"accuracy": accuracy_score(p.label_ids, preds),
            "f1_macro": f1_score(p.label_ids, preds, average="macro")}
```

```python
args = TrainingArguments(
    output_dir="out", eval_strategy="epoch", save_strategy="epoch",
    learning_rate=2e-5, per_device_train_batch_size=32,
    num_train_epochs=2, weight_decay=0.01, fp16=True,
    load_best_model_at_end=True, metric_for_best_model="f1_macro",
    report_to="none")

trainer = Trainer(model=model, args=args, train_dataset=train_t,
                  eval_dataset=val_t, processing_class=tok,
                  data_collator=DataCollatorWithPadding(tok),
                  compute_metrics=metrics)
trainer.train()
print(trainer.evaluate(test_t))
```

```python
from sklearn.metrics import classification_report, confusion_matrix

out = trainer.predict(test_t)
preds = np.argmax(out.predictions, axis=1)

print(classification_report(test["label"], preds,
      target_names=["negativo", "neutro", "positivo"]))
print(confusion_matrix(test["label"], preds))
```

```python
errors = [(t, l, p) for t, l, p in zip(test["review_text"], test["label"], preds) if l != p]
print(len(errors), "erros de", len(preds))

for t, l, p in errors[:30]:
    print(f"real={id2label[l]} | previsto={id2label[p]} | {t[:150]}\n")
```
